# 8. Interview grilling — Problem formulation

### Q1. Why did you convert a recommendation dataset into classification?

**Answer:** The original H&M problem is recommendation, but I wanted to demonstrate a broad core-ML workflow. Customer purchase propensity is a natural supervised-learning formulation and lets me study classification, imbalance, temporal leakage, model comparison, ensembles and threshold optimization.

### Q2. What exactly is your target?

**Answer:** For a cutoff date `t`, the target is 1 if the customer makes at least one purchase during `(t, t + 30 days]`; otherwise it is 0.

### Q3. Why 30 days?

**Answer:** Thirty days provides a concrete business prediction horizon and makes the target interpretable as short-term purchase propensity.

### Q4. Why is this a binary classification problem?

**Answer:** The output is a binary event: purchase versus no purchase during the future 30-day window.

### Q5. What would be the alternative formulation?

**Answer:** We could formulate it as recommendation, purchase-frequency regression, time-to-event prediction, or multi-class prediction based on future purchase frequency.

# 9. Interview grilling — Data leakage

### Q6. What is the biggest leakage risk?

**Answer:** Using transactions occurring after the feature cutoff to construct customer features. For example, calculating total spending using the entire transaction history would expose the model to future behavior.

### Q7. How did you prevent it?

**Answer:** For every temporal cutoff, features were computed only from transactions available up to that cutoff. The target was constructed separately from the following 30-day window.

### Q8. Why not randomly split the customers?

**Answer:** Random customer splitting can allow the model to learn patterns from future time periods while being evaluated on earlier periods. The real deployment scenario is predicting future behavior, so validation should mimic that.

### Q9. Why isn't random K-Fold appropriate?

**Answer:** Random K-Fold ignores temporal ordering. In a forecasting-like setting, that can make validation unrealistically optimistic.

### Q10. Why is preprocessing leakage possible?

**Answer:** If an imputer, scaler or encoder is fitted using validation/test information, statistics or category information from the future can influence training. Therefore preprocessing must be fitted only on training data.

# 10. Interview grilling — Feature engineering

### Q11. Why RFM?

**Answer:** Recency captures how recently a customer engaged, frequency captures purchasing activity, and monetary value captures spending. Together they summarize customer engagement and value.

### Q12. Why is recency useful?

**Answer:** A customer who purchased recently may be more likely to purchase again than a customer whose last purchase was much earlier.

### Q13. Why use recent 30-day and 90-day features in addition to lifetime features?

**Answer:** Lifetime features describe long-term behavior, while recent windows capture current momentum and changing customer intent.

### Q14. Why log-transform spending and item counts?

**Answer:** These variables are typically heavily right-skewed. `log1p` compresses extreme values and can make the feature distribution easier for some models to learn.

### Q15. Why create missing indicators?

**Answer:** Missingness can itself contain information. An indicator lets the model distinguish an actual value from a missing value that was later imputed.

# 11. Interview grilling — Metrics

### Q16. Why PR-AUC as the primary metric?

**Answer:** The positive class can be imbalanced. PR-AUC focuses on the precision-recall trade-off for the positive class and is often more informative than accuracy in such settings.

### Q17. Why not accuracy?

**Answer:** If most customers do not purchase, a model predicting no purchase for everyone can have high accuracy while being useless for identifying likely purchasers.

### Q18. Precision vs recall?

**Answer:** Precision answers: among customers predicted to purchase, how many actually purchase? Recall answers: among actual purchasers, how many did we identify?

### Q19. When would you prefer high recall?

**Answer:** If missing a likely purchaser is expensive, for example when the business wants to maximize the number of potential purchasers reached.

### Q20. When would you prefer high precision?

**Answer:** If contacting customers has a high cost, we may want to target fewer customers but make those targets more reliable.

# 12. Interview grilling — Bias and variance

### Q21. What is high bias?

**Answer:** The model is too simple and systematically underfits the underlying relationship.

### Q22. What is high variance?

**Answer:** The model fits training data very well but performs substantially worse on unseen data.

### Q23. How did you investigate this?

**Answer:** I compared training and validation performance and examined the generalization gap. I also compared models of different complexity.

### Q24. How can Random Forest reduce variance?

**Answer:** It trains many trees on bootstrapped samples and aggregates their predictions, reducing dependence on any single high-variance tree.

### Q25. Why can a deep decision tree overfit?

**Answer:** A deep tree can create highly specific decision boundaries that fit noise and individual training observations.

# 13. Interview grilling — Ensembles

### Q26. Bagging vs boosting?

**Answer:** Bagging trains models independently, usually on resampled datasets, and aggregates them to reduce variance. Boosting trains models sequentially, with later models focusing on errors made by earlier models.

### Q27. Random Forest vs Extra Trees?

**Answer:** Both are tree ensembles. Extra Trees introduce more randomness into split selection, while Random Forest generally searches for a stronger split among randomly selected features.

### Q28. Why use stacking?

**Answer:** Stacking allows a meta-model to learn how to combine predictions from different base learners instead of assigning fixed equal weights.

### Q29. Why can ensembles improve performance?

**Answer:** If individual models make partially different errors, combining them can reduce overall prediction error.

### Q30. Why measure prediction correlation?

**Answer:** Highly correlated models often make similar predictions. Models with complementary predictions can provide more benefit when combined.

# 14. Interview grilling — Gradient methods

### Q31. What is gradient descent?

**Answer:** An iterative optimization algorithm that updates model parameters in the direction that decreases the objective function.

### Q32. Batch vs SGD vs mini-batch?

**Answer:**
- Batch GD uses the entire training dataset per update.
- SGD uses one sample per update.
- Mini-batch GD uses a small subset per update.

### Q33. Why use mini-batches?

**Answer:** They provide a balance between the noisy updates of SGD and the computational cost of full-batch gradient descent, and can exploit vectorized hardware efficiently.

### Q34. What does the learning rate control?

**Answer:** It controls the size of parameter updates. Too small can make optimization slow; too large can cause instability or overshooting.

# 15. Interview grilling — PCA

### Q35. Why use PCA?

**Answer:** PCA transforms correlated numerical variables into orthogonal components ordered by explained variance. It can reduce dimensionality and help visualize or compress continuous features.

### Q36. Why not apply ordinary PCA directly to the complete one-hot encoded matrix?

**Answer:** One-hot encoding can produce a very large sparse matrix. Standard PCA generally requires dense data, which can create severe memory problems. Therefore PCA was applied to continuous numerical features rather than blindly densifying the complete sparse matrix.

### Q37. Does PCA always improve model accuracy?

**Answer:** No. PCA is a dimensionality-reduction technique, not a guaranteed performance improvement. It can remove noise or redundancy, but it can also discard predictive information.

# 16. Interview grilling — Threshold optimization

### Q38. Why isn't 0.5 always the correct threshold?

**Answer:** The probability threshold determines the precision-recall trade-off. With class imbalance or asymmetric business costs, another threshold may be more appropriate.

### Q39. How did you choose the threshold?

**Answer:** I evaluated a range of thresholds on the validation set and selected the threshold according to the chosen operating objective, with F1 used as the default optimization criterion in Phase 12.

### Q40. Why not optimize the threshold on test data?

**Answer:** That would make the test set part of model selection and produce an optimistic estimate of generalization.

# 17. Interview grilling — Error analysis

### Q41. What is a false positive?

**Answer:** The model predicts that the customer will purchase, but the customer does not purchase in the future 30-day window.

### Q42. What is a false negative?

**Answer:** The customer actually purchases, but the model predicts no purchase.

### Q43. Why analyze false negatives?

**Answer:** They reveal customers the model fails to identify and can expose missing behavioral signals or weak regions of the feature space.

### Q44. Why analyze performance by customer segment?

**Answer:** Overall metrics can hide systematic weaknesses. Segment analysis can show whether performance differs across age groups, engagement levels or other customer categories.

# 18. Interview grilling — Production

### Q45. Why save the preprocessing pipeline?

**Answer:** The model expects data in exactly the same transformed representation used during training. Saving the preprocessor ensures inference uses the same imputation, encoding and scaling logic.

### Q46. Why save the feature schema?

**Answer:** It allows the inference system to detect missing or unexpected inputs before producing predictions.

### Q47. Batch inference vs real-time inference?

**Answer:** Batch inference predicts many customers periodically and is suitable for campaigns or daily customer scoring. Real-time inference predicts an individual customer on demand and requires a low-latency API.

### Q48. What would you monitor after deployment?

**Answer:** Prediction volume, latency, missingness, feature drift, prediction distribution, class distribution when labels arrive, calibration, PR-AUC/recall/precision, and business outcomes.

# 19. Hard interview questions

### Q49. Your test PR-AUC improved. How do you know it isn't overfitting?

**Answer:** Model selection and hyperparameter tuning were performed without using the final test set. The test period was kept as a holdout and evaluated only after the model and threshold were frozen.

### Q50. Your validation score is much higher than test score. Why?

**Answer:** Possible reasons include temporal distribution shift, changing customer behavior, different class prevalence, overfitting to the validation period, or a mismatch between validation and deployment conditions.

### Q51. Why not use the customer's ID as a feature?

**Answer:** Customer ID is an identifier rather than meaningful behavioral information. Using it can lead to memorization and poor generalization.

### Q52. Why not delete outliers?

**Answer:** In customer behavior data, extreme values may represent genuine high-value customers rather than errors. I therefore identified outliers and used transformations/robust analysis rather than automatically deleting them.

### Q53. What is the biggest limitation of this project?

**Answer:** The target only asks whether at least one purchase occurs within 30 days. It does not predict which product the customer will purchase, how many products they will buy, or the exact purchase date.

### Q54. What would you improve next?

**Answer:** I would investigate probability calibration, cost-sensitive threshold selection, richer sequential customer features, temporal drift monitoring, and potentially reformulate the problem as ranking/recommendation or time-to-event prediction.
